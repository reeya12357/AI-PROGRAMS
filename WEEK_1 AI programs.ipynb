```python


board = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]



def display_board():
    print()
    for i in range(3):
        print(" | ".join(board[i]))
        if i < 2:
            print("---------")
    print()



def check_win(player):
    # Check rows
    for i in range(3):
        if (board[i][0] == player and
            board[i][1] == player and
            board[i][2] == player):
            return True


    for i in range(3):
        if (board[0][i] == player and
            board[1][i] == player and
            board[2][i] == player):
            return True


    if (board[0][0] == player and
        board[1][1] == player and
        board[2][2] == player):
        return True

    if (board[0][2] == player and
        board[1][1] == player and
        board[2][0] == player):
        return True

    return False



def board_full():
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                return False
    return True



print("Tic-Tac-Toe")
print("Enter row and column numbers from 1 to 3.")

display_board()

player = "X"

while True:

    print(f"Player {player}, enter row: ", end="")
    row = int(input())

    print(f"Player {player}, enter column: ", end="")
    column = int(input())


    row = row - 1
    column = column - 1


    if row < 0 or row > 2 or column < 0 or column > 2:
        print("Invalid position! Enter numbers from 1 to 3.")
        continue


    if board[row][column] != " ":
        print("That position is already occupied!")
        continue


    board[row][column] = player

    display_board()


    if check_win(player):
        print(f"Player {player} wins!")
        break


    if board_full():
        print("It's a draw!")
        break


    if player == "X":
        player = "O"
    else:
        player = "X"
```

```python
class VacuumCleaner:
    def __init__(self, location, room_a, room_b):
        self.location = location
        self.rooms = {
            "A": room_a,
            "B": room_b
        }

    def display_state(self):
        print("\nCurrent State:")
        print(f"Vacuum Cleaner Location : Room {self.location}")
        print(f"Room A                 : {self.rooms['A']}")
        print(f"Room B                 : {self.rooms['B']}")

    def clean(self):
        """Clean the current room if it is dirty."""
        if self.rooms[self.location] == "Dirty":
            print(f"Action: SUCK -> Cleaning Room {self.location}")
            self.rooms[self.location] = "Clean"
        else:
            print(f"Room {self.location} is already clean.")

    def move_left(self):
        """Move from B to A."""
        if self.location == "B":
            print("Action: MOVE LEFT -> Room A")
            self.location = "A"
        else:
            print("Vacuum is already in Room A.")

    def move_right(self):
        """Move from A to B."""
        if self.location == "A":
            print("Action: MOVE RIGHT -> Room B")
            self.location = "B"
        else:
            print("Vacuum is already in Room B.")

    def is_goal_reached(self):
        """Check whether both rooms are clean."""
        return (
            self.rooms["A"] == "Clean"
            and self.rooms["B"] == "Clean"
        )

    def run(self):
        """Run the vacuum cleaner agent."""
        print("\n=== Vacuum Cleaner Agent ===")

        self.display_state()


        while not self.is_goal_reached():


            self.clean()


            if self.location == "A":
                self.move_right()
            else:
                self.move_left()

            self.display_state()

        print("\nGoal Reached!")
        print("Both Room A and Room B are CLEAN.")



agent = VacuumCleaner(
    location="A",
    room_a="Dirty",
    room_b="Dirty"
)

agent.run()
```
